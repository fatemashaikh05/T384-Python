### Concatenation of Arrays

```python
A1 = np.ones((3,5), dtype=int) 
A2 = np.zeros((3,2), dtype=int) 
```

```python
print(A1)
print(A2)
```

```python
A3= np.concatenate((A1,A2),axis=1)
A3
```

```python
A1 = np.ones((3,5), dtype=int) 
A2 = np.zeros((2,5), dtype=int) 
```

```python
print(A1)
print(A2)
```

```python
A3= np.concatenate((A1,A2),axis=0)
A3
```

### Concatenation using : HStack and Vstack

```python
A1 = np.ones((3,5), dtype=int) 
A2 = np.zeros((3,5), dtype=int) 

np.hstack((A1,A2))
```

```python
np.vstack((A1,A2))
```

```python
import numpy as np   # Linearly equally spaced numbers
arr1 = np.linspace(1,100,18).reshape(2,3,3)
arr1
```

```python
# Slicing
arr_slice = arr1[1,:,:] # arr1[matrices,row,column]
arr_slice
```

```python
# Flattening
# it reduces dimensions to one dimension.
print(arr_slice)
print("Dimension of ORIGINAL array is:", arr_slice.ndim)
print("------------------------------------------")
print(arr_slice.flatten())
print("Dimension of FLATTENED array is:", arr_slice.flatten().ndim)
```

```python
print(arr_slice)
print("Dimension of ORIGINAL array is:", arr_slice.ndim)
print("------------------------------------------")
print(arr_slice.ravel())
print("Dimension of FLATTENED array is:", arr_slice.ravel().ndim)
```
