# lambda Functions

```python
num_square = lambda x :x**2
print(num_square(3))
```

```python
greetings = lambda : "Hello python"
print(greetings())
```

```python
add = lambda a: a+300
print(add(250))
```

```python
num = lambda x, y :x+y
print(num(3,5))
```

```python
num = lambda a, b :a*b
print(num(3,5))
```

```python
maximum = lambda a, b: a if a > b else b
print(maximum(50, 20))
```

```python
def remainder(x,y):
    return x%y
remainder(45,6)
```

```python
check_even = lambda x: "Even" if x%2==0 else "Odd"
print(check_even(10))
```

```python
data = lambda x: "Even" if x % 2 == 0 else "Odd"
```

```python
list1 = [4, 6, 7, 2, 9]
result = [ (lambda x: "Even" if x % 2 == 0 else "Odd")(x) for x in list1 ]

print(result)
```

```python
list = [4,6,7,2,9]

for num in list:
    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")
```

```python
amount = int(input("Enter amount: "))

n100 = 0
n50 = 0
n10 = 0

while amount>=100 and n100<40:
    amount = amount - 100
    n100 = n100 + 1

while amount>=50 and n50<20:
    amount = amount - 50
    n50 = n50 + 1

while amount>=10 and n10<100:
    amount = amount - 10
    n10 = n10 + 1

if amount == 0:
    print("100 rupees notes:", n100)
    print("50 rupees notes:", n50)
    print("10 rupees notes:", n10)
else:
    print("Amount cannot be removed!")
```

```python
amount = int(input("Enter amount to withdraw: "))

notes100 = 40
notes50 = 20
notes10 = 100

if amount % 10 != 0:
    print("Amount should be a multiple of 10")
else:
    n100 = min(amount // 100, notes100)
    amount = amount - n100 * 100

    n50 = min(amount // 50, notes50)
    amount = amount - n50 * 50

    n10 = min(amount // 10, notes10)
    amount = amount - n10 * 10

    if amount == 0:
        print("100 rupees notes:", n100)
        print("50 rupees notes:", n50)
        print("10 rupees notes:", n10)
    else:
        print("Amount cannot be removed")
```

```python
amount = int(input("Enter amount to withdraw: "))

n100 = amount // 100
if n100 > 40:
    n100 = 40
amount = amount - n100 * 100

n50 = amount // 50
if n50 > 20:
    n50 = 20
amount = amount - n50 * 50

n10 = amount // 10
if n10 > 100:
    n10 = 100
amount = amount - n10 * 10

if amount == 0:
    print("100 rupees notes:", n100)
    print("50 rupees notes:", n50)
    print("10 rupees notes:", n10)
else:
    print("Amount cannot be dispensed")
```
