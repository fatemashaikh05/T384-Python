```python
def multiply(*args):
    result = 1
    for i in args:
        result = result * i
    return result

# ----------------------------------

nos = int(input("Tell me how many numbers you want to multiply? "))
list_nos = []

for i in range(nos):
    num = int(input("Enter number: "))
    list_nos.append(num)

print("Multiplication:", multiply(*list_nos))
```

# **KWARGS

```python
dict1 = {"ID":1001,"Name":"John Doe","Department":"IT"}
```

```python
dict1 = {"ID":1001,"Name":"John Doe","Department":"IT"}

for i in dict1.keys():
    print(i)
```

```python
dict1 = {"ID":1001,"Name":"John Doe","Department":"IT"}

for key, value in dict1.items():
    print(key, value)
```

```python
def details(**kwargs):
    for key, value in kwargs.items():
        print(key,value)

dict1 = {"ID":1001,"Name":"John Doe","Department":"IT","Salary":80000}
details(**dict1)
```

```python
def details(**india):
    for key, value in india.items():
        print(key,value)

dict1 = {"ID":1001,"Name":"John Doe","Department":"IT","Salary":80000}
details(**dict1)
```

```python
def details(**india):
    print("Details of your Stay: \n--------------------")
    for key, value in india.items():
        print(f"{key}: {value}")

Hotel_details = {"City":"Mumbai","Location":"Juhu","Hotel":"Sky","Room":12}

details(**Hotel_details)
```
