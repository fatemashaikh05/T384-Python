# Exception Handling

```python
x=100
try:
    print(200/x)
    
except ZeroDivisionError:
    print("Number was divided by ZERO!")
    
except Exception:
    print("Error occured")

finally:
    print("Program Ends!")
```

```python
x=100
try:
    print(200/x)

except NameError:
    print("Wrong Keyword spelling, plz check!")
    
except ZeroDivisionError:
    print("Number was divided by ZERO!")
    
except Exception:
    print("Error occured")

finally:
    print("Program Ends!")
```

```python
try:
    num = int(input("Enter Numbers Only"))
    
except ValueError:
    print("Hello Sir/Madam, we request you to enter number only..")
    try:
        num = int(input("Enter Numbers Only(Last chance to attempt)"))
    except ValueError:
        print("Sorry! You had not entered the numbers, Please remove your card and try again")
finally:
    print("Thankyou ! Visit Again.")
```
