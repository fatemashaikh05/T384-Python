```python
# WAP to find factorial of a num in while loop
n = int(input("Enter a number: "))

fact = 1
while n > 0:
    fact = fact * n
    n = n - 1

print("Factorial =", fact)
```

```python
# WAP to find factorial of a number using a for loop
n = int(input("Enter a number: "))

fact = 1
for i in range(1, n + 1):
    fact = fact * i

print("Factorial = ", fact)
```

```python
# WAP to print numbers divisible by both 7 and 9
# Loop from 1 to 100
for i in range(1, 101):
    # Check if i is divisible by both 7 and 9
    if i % 7 == 0 and i % 9 == 0:
        print("Number =", i)
        # Stop the loop
        break
```

```python
# WAP to print numbers from 1 to 100
# Stop when a number is divisible by both 7 and 9
for i in range(1, 101):
    # Print the current number
    print(i)
    # Check if the number is divisible by both 7 and 9
    if i % 7 == 0 and i % 9 == 0:
        # Stop the loop
        break
```

```python
# WAP to print numbers from 1 to 100
# Skip numbers divisible by 7
for i in range(1, 101):
    # Check if the number is divisible by 7
    if i % 7 == 0:
        continue

    # Print the number
    print(i)
```

```python
# WAP to print numbers from 1 to 100
# Skip numbers divisible by 7
i = 1

while i <= 100:
    # Check if the number is divisible by 7
    if i % 7 == 0:
        i = i + 1
        continue
    print(i)
    i = i + 1
```
