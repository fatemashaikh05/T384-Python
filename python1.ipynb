# Type Casting

```python
A="5"
print(A)
print(type(A))

# converting string to integer
A=int(A)
print(A)
print(type(A))

# converting string to float
A=float(A)
print(A)
print(type(A))

# converting string to complex
A=complex(A)
print(A)
print(type(A))

# converting string to boolean
A=bool(A)
print(A)
print(type(A))
```
