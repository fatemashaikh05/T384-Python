```python
class club_Mahindra:
    # Constructor
    def __init__(self, cust_ID, Name, Membership, Subscription):
        self.Customer_ID = cust_ID
        self.Customer_Name = Name
        self.Membership_Info = Membership
        self.Subs = Subscription

    # Information method
    def information(self):
        print(f"Welcome {self.Customer_Name}")
        print("Customer ID:", self.Customer_ID)
        print("Membership:", self.Membership_Info)
        print("Subscription:", self.Subs)

# Creating an object
c1 = club_Mahindra(101, "Rahul", "Gold", "Annual")

# Calling the method
c1.information()
```

```python
# Parent class
class club_Mahindra:
    def __init__(self, cust_ID, Name, Membership, Subscription):
        self.Customer_ID = cust_ID
        self.Customer_Name = Name
        self.Membership_Info = Membership
        self.Subs = Subscription

    def information(self):
        print(f"Welcome {self.Customer_Name}")
        print("Customer ID:", self.Customer_ID)
        print("Membership:", self.Membership_Info)
        print("Subscription:", self.Subs)

# Child class
class new_requirement(club_Mahindra):
    def __init__(self, cust_ID, Name, Membership, Subscription, Aadhaar):
        super().__init__(cust_ID, Name, Membership, Subscription)
        self.Aadhaar_Number = Aadhaar

    def information(self):
        super().information()
        print("Aadhaar Number:", self.Aadhaar_Number)

# Creating objects
obj1 = new_requirement(1001, "Rahul", "Gold", "Annual", "XXXX-XXXX-1234")
obj2 = new_requirement(1002, "Priya", "Platinum", "Yearly", "XXXX-XXXX-5678")
obj3 = new_requirement(1003, "Jaya", "Diamond", "Monthly", "XXXX-XXXX-9012")

obj3.information()
```

```python
class Vehicle:
    def vehicle_info(self):
        print("Inside vehicle class")

class Car(Vehicle):
    def car_info(self):
        print("Inside car class")

# object
c1 = Car()
# calling
c1.car_info()
c1.vehicle_info()
```

# Multilevel Inheritance

```python
class Vehicle:
    def vehicle_info(self):
        print("Inside vehicle class")

class Car(Vehicle):
    def car_info(self):
        print("Inside car class")

class SportsCar(Car):
    def SportsCar_info(self):
        print("Inside Sports Car class")
        
# object
s1 = SportsCar()
# calling
s1.SportsCar_info()
s1.car_info()
s1.vehicle_info()
```

```python
class Jar:
    def __init__(self):
        self.candies = 100


class Withdraw:
    def execute(self, jar, amount):
        if amount <= jar.candies:
            jar.candies -= amount
            print("Withdraw:", amount)
        else:
            print("Not enough candies!")

        print("Remaining candies:", jar.candies)


class Deposit:
    def execute(self, jar, amount):
        jar.candies += amount
        print("Deposit:", amount)
        print("Total candies:", jar.candies)


jar = Jar()

user1 = Withdraw()
user2 = Deposit()
user3 = Withdraw()
user4 = Deposit()

amount = int(input("User 1: Enter candies to withdraw: "))
user1.execute(jar, amount)

amount = int(input("User 2: Enter candies to deposit: "))
user2.execute(jar, amount)

amount = int(input("User 3: Enter candies to withdraw: "))
user3.execute(jar, amount)

amount = int(input("User 4: Enter candies to deposit: "))
user4.execute(jar, amount)
```

```python
class Jar:
    def __init__(self):
        self.capacity = 100
        self.candies = 100

        print("Jar Capacity:", self.capacity)
        print("Initial Candies:", self.candies)


class Withdraw:
    def execute(self, jar, candies_count):
        if candies_count <= 0:
            print("Enter a valid number of candies!")
        elif candies_count <= jar.candies:
            jar.candies -= candies_count
            print("Candies withdrawn:", candies_count)
        else:
            print("Not enough candies!")

        print("Remaining candies:", jar.candies)


class Deposit:
    def execute(self, jar, candies_count):
        if candies_count <= 0:
            print("Enter a valid number of candies!")
        elif jar.candies + candies_count <= jar.capacity:
            jar.candies += candies_count
            print("Candies deposited:", candies_count)
        else:
            print("Jar capacity exceeded!")

        print("Remaining candies:", jar.candies)


jar = Jar()

user1 = Withdraw()
user2 = Deposit()
user3 = Withdraw()
user4 = Deposit()

candies_count = int(input("User 1: Enter candies to withdraw: "))
user1.execute(jar, candies_count)

candies_count = int(input("User 2: Enter candies to deposit: "))
user2.execute(jar, candies_count)

candies_count = int(input("User 3: Enter candies to withdraw: "))
user3.execute(jar, candies_count)

candies_count = int(input("User 4: Enter candies to deposit: "))
user4.execute(jar, candies_count)

print("\nFinal candies in jar:", jar.candies)
```
