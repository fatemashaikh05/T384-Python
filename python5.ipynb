# Recursive Function

```python
# WAP to print Factorial of a num using recursive function
def fact(n):
    if n==1:
        return 1
    else:
        return n * fact(n - 1)  # Recursive call

num = int(input("Enter a number: "))

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    print("Factorial =", fact(num))
```

```python
def fact(n):
    if n==1:
        return 1
    else:
        return n * fact(n - 1)  # Recursive call
fact(4)
```

```python
# WAP to print Fibonacci series using recursive function
def fib(n):
    if n<=1:
        return n
    else:
        return fib(n-1)+fib(n-2)

num = int(input("Enter a number: "))

for i in range(num):
    print(fib(i), end=" ")
```

```python
# WAP to print Fibonacci series using for loop
num=int(input("Enter a number:"))

a=0
b=1

for i in range(num):
    print(a, end=" ")
    c=a+b
    a=b
    b=c
```

```python
# WAP to print Fibonacci series using while loop
num = int(input("Enter the number of terms: "))

a = 0
b = 1
i = 0

while i < num:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    i += 1
```
